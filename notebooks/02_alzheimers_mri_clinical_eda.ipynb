# NeuroTrace: Alzheimer's & Dementia Clinical & MRI Volumetric EDA

Exploratory data analysis of longitudinal cognitive scores (MMSE, CDR) and structural MRI volumetric measures (nWBV, eTIV) from the Washington University OASIS cohort.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
%matplotlib inline

df = pd.read_csv('../data/raw/oasis_longitudinal.csv')
print(f'OASIS Cohort Shape: {df.shape}')
df.head()
```

### 1. Group Distribution & Missing Values

```python
print(df['Group'].value_counts())
print('Missing Values:')
print(df.isnull().sum()[df.isnull().sum() > 0])
```

### 2. Normalized Whole Brain Volume (nWBV) vs Age
Brain atrophy progression across diagnostic groups:

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='nWBV', hue='Group', style='Group', s=70, alpha=0.8)
plt.title('Normalized Whole Brain Volume (nWBV) vs Age')
plt.xlabel('Age (Years)')
plt.ylabel('nWBV (Brain Parenchymal Ratio)')
plt.show()
```

### 3. Mini-Mental State Examination (MMSE) by CDR

```python
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x='CDR', y='MMSE', palette='Greens')
plt.title('MMSE Cognitive Score across Clinical Dementia Ratings (CDR)')
plt.show()
```
