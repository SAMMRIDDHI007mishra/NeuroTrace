# NeuroTrace: Parkinson's Vocal Acoustic Biomarker EDA

Exploratory data analysis of fundamental frequency, jitter, shimmer, and dynamical complexity measures from the UCI Machine Learning Repository (Little et al., 2007).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
%matplotlib inline

df = pd.read_csv('../data/raw/parkinsons.data')
print(f'Dataset Shape: {df.shape}')
df.head()
```

### 1. Target Class Balance
Distribution of Healthy Controls (0) vs Confirmed Parkinson's Disease (1):

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='status', palette='Blues')
plt.title('Class Distribution: Healthy (0) vs Parkinson\'s (1)')
plt.xlabel('Clinical Status')
plt.ylabel('Count')
plt.show()

print(df['status'].value_counts(normalize=True))
```

### 2. Acoustic Biomarker Distributions
Comparing Pitch Period Entropy (PPE), Spread1, and Harmonics-to-Noise Ratio (HNR):

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.boxplot(ax=axes[0], data=df, x='status', y='PPE', palette='Blues')
axes[0].set_title('Pitch Period Entropy (PPE)')

sns.boxplot(ax=axes[1], data=df, x='status', y='HNR', palette='Blues')
axes[1].set_title('Harmonics-to-Noise Ratio (HNR)')

sns.boxplot(ax=axes[2], data=df, x='status', y='spread1', palette='Blues')
axes[2].set_title('Non-linear Spread 1')
plt.tight_layout()
plt.show()
```

### 3. Correlation Matrix Heatmap

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr = df[numeric_cols].corr()

plt.figure(figsize=(14, 10))
sns.heatmap(corr, cmap='coolwarm', center=0, annot=False)
plt.title('Correlation Heatmap of Vocal Acoustic Features')
plt.show()
```
